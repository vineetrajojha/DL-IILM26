Method 1: Download from kaggle.

```python
#importing the dataset from kaggle
import kagglehub
import pandas as pd
import os

# download dataset
path = kagglehub.dataset_download("uciml/iris")

# file path
print(os.listdir(path))

# load file
csv_file = [f for f in os.listdir(path) if f.endswith(".csv") or f.endswith(".data")][0]
file_path = os.path.join(path, csv_file)

df = pd.read_csv(file_path)

# header
print(df.columns)

# head
print(df.head())

# tail
print(df.tail())
```
